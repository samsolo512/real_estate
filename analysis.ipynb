```python
import pandas as pd
from IPython.display import HTML, display
```

```python
# df = pd.read_csv('herriman_investment_analysis.csv')
df = pd.read_csv('herriman_homes.csv')
```

```python
def show_df(df, rows=1000, cols=100, max_height='400px'):
    # Convert DataFrame to HTML
    html = df.head(rows).to_html()
    
    # Add CSS to make rows single-line height
    styled_html = f"""
    <div style="height: {max_height}; width: 100%; overflow: auto; border: 1px solid #ddd; padding: 10px;">
        <style>
            table {{
                border-collapse: collapse;
                width: 100%;
            }}
            td, th {{
                border: 1px solid #ddd;
                padding: 4px 8px;
                text-align: left;
                white-space: nowrap;
                overflow: hidden;
                text-overflow: ellipsis;
                max-width: 200px;
                height: 20px;
                line-height: 20px;
            }}
            tr {{
                height: 20px;
            }}
        </style>
        {html}
    </div>
    """
    display(HTML(styled_html))
```

```python
show_df(df)
```
